**Customer** **Reviews** **Table** **Cleaning**

**Transaction** **Raw** **Table** **cleaning**

```python
transaction_raw=pd.read_csv("/content/drive/MyDrive/CoffeeBites_Project1_Starter_Dataset/transactions_raw.csv")
transaction_raw
```

```python
transaction_raw.info()
```

```python
#order_datetime Dtype changing
transaction_raw['order_datetime']=pd.to_datetime(transaction_raw['order_datetime'],format="mixed",errors='coerce')
```

```python
transaction_raw.shape
```

```python
transaction_raw.describe()
```

```python
#Gap filling
text_col2=['customer_id']
for i in text_col2:
  transaction_raw[i]=transaction_raw[i].fillna("Unknown")
```

```python
#order_date  gap filling
transaction_raw['order_datetime'].isna().sum()
```

```python
transaction_raw.isnull().sum()
```

```python
transaction_raw['payment_method']=transaction_raw['payment_method'].str.strip().str.title()
```

```python
transaction_raw['payment_method'].value_counts()
```

```python
#Detect duplicates and remove
transaction_raw_dup=transaction_raw[transaction_raw['transaction_id'].duplicated(keep=False)].sort_values('transaction_id')
transaction_raw_dup
```

```python
transaction_raw_clean=transaction_raw.drop_duplicates('transaction_id',keep="first")
transaction_raw_clean
```

```python
print("Before cleaning:",len(transaction_raw))
print("After Cleaning:",len(transaction_raw_clean))
```

**Referential** **integrity** **checks**

```python
invalid_products = transaction_items_clean[
    ~transaction_items_clean["product_id"].isin(
        product["product_id"]
    )
]

print("Invalid product references:", len(invalid_products))
```

```python
print(invalid_products["product_id"].unique())
```

```python
invalid_transaction = transaction_items_clean[
    ~transaction_items_clean["transaction_id"].isin(
        transaction_raw_clean["transaction_id"]
    )
]

print("Invalid transaction references:", len(invalid_transaction))
```

```python
invalid_customers = transaction_raw_clean[
    transaction_raw_clean["customer_id"].notna()
    & (transaction_raw_clean["customer_id"] != "Unknown")
    & ~transaction_raw_clean["customer_id"].isin(
        customers["customer_id"]
    )
]

print("Invalid customer references:", len(invalid_customers))
```

```python
print("Duplicate transaction IDs:",
      transaction_raw_clean["transaction_id"].duplicated().sum())

print("Invalid product references:", len(invalid_products))

print("Invalid transaction references:", len(invalid_transaction))

print("Invalid customer references:", len(invalid_customers))
```

```python
# Export cleaned tables as CSV files

product.to_csv("Product.csv", index=False)
customers.to_csv("Customers.csv", index=False)
transaction_raw_clean.to_csv("Transaction_raw_Clean.csv", index=False)
transaction_items_clean.to_csv("Transaction_Items_Clean.csv", index=False)
customer_reviews_clean.to_csv("Customer_Reviews_Clean.csv", index=False)

print("All cleaned CSV files exported successfully!")
```
